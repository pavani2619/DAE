```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# 2. LOAD DATASET
from google.colab import files
uploaded = files.upload()
# Read uploaded CSV file automatically
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)
print("COVID-19 DATASET LOADED SUCCESSFULLY")
print("=" * 50)
# 3. DATASET INFORMATION
print("\nFIRST 5 RECORDS:")
print(df.head())
print("\nDATASET SHAPE:")
print(df.shape)
print("\nCOLUMN NAMES:")
print(df.columns.tolist())
print("\nDATASET INFORMATION")
# 4. CHECK MISSING VALUES
print("\nMISSING VALUES:")
print(df.isnull().sum())
total_missing = df.isnull().sum().sum()
print("\nTOTAL MISSING VALUES:", total_missing)
# 5. CHECK DUPLICATE RECORDS
duplicate_count = df.duplicated().sum()
print("\nDUPLICATE RECORDS:", duplicate_count)
# 6. REMOVE DUPLICATE RECORDS
df = df.drop_duplicates()
print("\nAFTER REMOVING DUPLICATES:")
print("Dataset Shape:", df.shape)
# 7. BASIC STATISTICS
print("\nSTATISTICAL SUMMARY:")
print(df.describe())
# 8. TOTAL COVID-19 STATISTICS
total_confirmed = df["Confirmed"].sum()
total_deaths = df["Deaths"].sum()
total_recovered = df["Recovered"].sum()
total_active = df["Active"].sum()
total_new_cases = df["New cases"].sum()
total_new_deaths = df["New deaths"].sum()
total_new_recovered = df["New recovered"].sum()
print("\n" + "=" * 60)
print("OVERALL COVID-19 STATISTICS")
print("=" * 60)
print("Total Confirmed Cases :", total_confirmed)
print("Total Deaths         :", total_deaths)
print("Total Recovered      :", total_recovered)
print("Total Active Cases   :", total_active)
print("\nNew Cases            :", total_new_cases)
print("New Deaths           :", total_new_deaths)
print("New Recovered        :", total_new_recovered)
# 9. COUNTRY WITH HIGHEST CONFIRMED CASES
highest_confirmed = df.loc[df["Confirmed"].idxmax()]
print("\n" + "=" * 60)
print("HIGHEST CONFIRMED CASES")
print("=" * 60)
print("Country :", highest_confirmed["Country/Region"])
print("Cases   :", highest_confirmed["Confirmed"])
# 10. COUNTRY WITH HIGHEST DEATHS
highest_deaths = df.loc[df["Deaths"].idxmax()]
print("\n" + "=" * 60)
print("HIGHEST DEATHS")
print("=" * 60)
print("Country :", highest_deaths["Country/Region"])
print("Deaths  :", highest_deaths["Deaths"])
# 11. COUNTRY WITH HIGHEST RECOVERIES
highest_recovered = df.loc[df["Recovered"].idxmax()]
print("\n" + "=" * 60)
print("HIGHEST RECOVERIES")
print("=" * 60)
print("Country  :", highest_recovered["Country/Region"])
print("Recovered:", highest_recovered["Recovered"])
# 12. COUNTRY WITH HIGHEST ACTIVE CASES
highest_active = df.loc[df["Active"].idxmax()]
print("\n" + "=" * 60)
print("HIGHEST ACTIVE CASES")
print("=" * 60)
print("Country :", highest_active["Country/Region"])
print("Active  :", highest_active["Active"])
# 13. TOP 10 COUNTRIES BY CONFIRMED CASES
top10_confirmed = df.sort_values(
    by="Confirmed",
    ascending=False
).head(10)

print("\nTOP 10 COUNTRIES BY CONFIRMED CASES:")
print(
    top10_confirmed[
        ["Country/Region", "Confirmed"]
    ].to_string(index=False)
)
# 14. BAR CHART - TOP 10 CONFIRMED CASES
plt.figure(figsize=(12, 6))

plt.bar(
    top10_confirmed["Country/Region"],
    top10_confirmed["Confirmed"]
)

plt.title("Top 10 Countries by Confirmed COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 15. TOP 10 COUNTRIES BY DEATH
top10_deaths = df.sort_values(
    by="Deaths",
    ascending=False
).head(10)
print("\nTOP 10 COUNTRIES BY DEATHS:")
print(
    top10_deaths[
        ["Country/Region", "Deaths"]
    ].to_string(index=False)
)
# 16. BAR CHART - TOP 10 DEATHS
plt.figure(figsize=(12, 6))
plt.bar(
    top10_deaths["Country/Region"],
    top10_deaths["Deaths"]
)
plt.title("Top 10 Countries by COVID-19 Deaths")
plt.xlabel("Country")
plt.ylabel("Deaths")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 17. TOP 10 COUNTRIES BY RECOVERED CASES
top10_recovered = df.sort_values(
    by="Recovered",
    ascending=False
).head(10)
print("\nTOP 10 COUNTRIES BY RECOVERED CASES:")
print(
    top10_recovered[
        ["Country/Region", "Recovered"]
    ].to_string(index=False)
)
# 18. BAR CHART - TOP 10 RECOVERED

plt.figure(figsize=(12, 6))
plt.bar(
    top10_recovered["Country/Region"],
    top10_recovered["Recovered"]
)
plt.title("Top 10 Countries by Recovered COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Recovered Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 19. TOP 10 COUNTRIES BY ACTIVE CASES
top10_active = df.sort_values(
    by="Active",
    ascending=False
).head(10)
print("\nTOP 10 COUNTRIES BY ACTIVE CASES:")
print(
    top10_active[
        ["Country/Region", "Active"]
    ].to_string(index=False)
)
# 20. BAR CHART - TOP 10 ACTIVE CASES
plt.figure(figsize=(12, 6))
plt.bar(
    top10_active["Country/Region"],
    top10_active["Active"]
)
plt.title("Top 10 Countries by Active COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Active Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 21. GROUPBY ANALYSIS
country_analysis = df.groupby("Country/Region")[
    ["Confirmed", "Deaths", "Recovered", "Active"]
].sum()
print("\nCOUNTRY-WISE GROUPBY ANALYSIS:")
print(country_analysis.head(10))
# 22. COMPARISON OF CONFIRMED, DEATHS AND RECOVERED
top5 = df.sort_values(
    by="Confirmed",
    ascending=False
).head(5)
x = np.arange(len(top5))
width = 0.25
plt.figure(figsize=(12, 6))
plt.bar(
    x - width,
    top5["Confirmed"],
    width,
    label="Confirmed"
)
plt.bar(
    x,
    top5["Deaths"],
    width,
    label="Deaths"
)

plt.bar(
    x + width,
    top5["Recovered"],
    width,
    label="Recovered"
)
plt.title("COVID-19 Case Comparison – Top 5 Countries")
plt.xlabel("Country")
plt.ylabel("Number of Cases")
plt.xticks(x, top5["Country/Region"], rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
# 23. DEATH RATE ANALYSIS
top10_death_rate = df.sort_values(
    by="Deaths / 100 Cases",
    ascending=False
).head(10)
print("\nTOP 10 COUNTRIES BY DEATH RATE:")
print(
    top10_death_rate[
        ["Country/Region", "Deaths / 100 Cases"]
    ].to_string(index=False)
)
# 24. DEATH RATES
plt.figure(figsize=(12, 6))
plt.bar(
    top10_death_rate["Country/Region"],
    top10_death_rate["Deaths / 100 Cases"]
)
plt.title("Top 10 Countries by Death Percentage")
plt.xlabel("Country")
plt.ylabel("Deaths per 100 Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 25. RECOVERY RATE ANALYSIS
top10_recovery_rate = df.sort_values(
    by="Recovered / 100 Cases",
    ascending=False
).head(10)
print("\nTOP 10 COUNTRIES BY RECOVERY RATE:")
print(
    top10_recovery_rate[
        ["Country/Region", "Recovered / 100 Cases"]
    ].to_string(index=False)
)
# 26. RECOVERY RATE GRAPH
plt.figure(figsize=(12, 6))

plt.bar(
    top10_recovery_rate["Country/Region"],
    top10_recovery_rate["Recovered / 100 Cases"]
)
plt.title("Top 10 Countries by Recovery Percentage")
plt.xlabel("Country")
plt.ylabel("Recovered per 100 Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
# 27. OVERALL COVID-19 CASE DISTRIBUTION
labels = [
    "Confirmed",
    "Deaths",
    "Recovered",
    "Active"
]
values = [
    total_confirmed,
    total_deaths,
    total_recovered,
    total_active
]
plt.figure(figsize=(8, 8))
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Overall COVID-19 Case Distribution")
plt.show()
# 28. DASHBOARD SUMMARY
print("\n")
print("=" * 70)
print("           COVID-19 DATA ANALYSIS DASHBOARD")
print("=" * 70)
print("\nNumber of Countries/Regions :", len(df))
print("Total Confirmed Cases       :", total_confirmed)
print("Total Deaths                :", total_deaths)
print("Total Recovered Cases       :", total_recovered)
print("Total Active Cases          :", total_active)
print("\nNew Cases                   :", total_new_cases)
print("New Deaths                  :", total_new_deaths)
print("New Recovered Cases         :", total_new_recovered)
print("\nHighest Confirmed Cases     :",
      highest_confirmed["Country/Region"])
print("Highest Deaths              :",
      highest_deaths["Country/Region"])
print("Highest Recoveries          :",
      highest_recovered["Country/Region"])
print("Highest Active Cases        :",
      highest_active["Country/Region"])
print("\n" + "=" * 70)
print("              ANALYSIS COMPLETED")
print("=" * 70)
```
