```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------
# Step 1 : Create Dataset

data = {
    "ApplicantIncome":[25000,30000,15000,40000,22000,
                       50000,18000,35000,27000,16000],

    "LoanAmount":[100,120,80,150,90,
                  180,70,140,110,75],

    "CreditHistory":[1,1,0,1,0,
                     1,0,1,1,0],

    "LoanStatus":["Approved","Approved","Rejected","Approved","Rejected",
                  "Approved","Rejected","Approved","Approved","Rejected"]
}

df = pd.DataFrame(data)

# ---------------------------------------
# Step 2 : Save Dataset as CSV
# ---------------------------------------

df.to_csv("loan_data.csv", index=False)

print("loan_data.csv created successfully.\n")

# ---------------------------------------
# Step 3 : Read CSV File
# ---------------------------------------

df = pd.read_csv("loan_data.csv")

print("Loan Dataset\n")
print(df)

# ---------------------------------------
# Step 4 : Dataset Information
# ---------------------------------------

print("\n==============================")
print("Dataset Information")
print("==============================")

print("\nShape of Dataset :", df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

# ---------------------------------------
# Step 5 : Statistical Analysis using NumPy
# ---------------------------------------

income = df["ApplicantIncome"].values
loan = df["LoanAmount"].values

print("\n==============================")
print("NUMPY ANALYSIS")
print("==============================")

print("Average Income :", np.mean(income))
print("Maximum Income :", np.max(income))
print("Minimum Income :", np.min(income))
print("Median Income :", np.median(income))
print("Standard Deviation :", np.std(income))

print("\nAverage Loan Amount :", np.mean(loan))
print("Maximum Loan :", np.max(loan))
print("Minimum Loan :", np.min(loan))

# ---------------------------------------
# Step 6 : Pandas Analysis
# ---------------------------------------

print("\n==============================")
print("PANDAS ANALYSIS")
print("==============================")

print(df.describe())

print("\nLoan Status Count")
print(df["LoanStatus"].value_counts())

print("\nCredit History Count")
print(df["CreditHistory"].value_counts())

# ---------------------------------------
# Step 7 : Histogram
# ---------------------------------------

plt.figure(figsize=(6,4))

plt.hist(df["ApplicantIncome"], bins=5)

plt.title("Applicant Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()

# ---------------------------------------
# Step 8 : Loan Approval Bar Chart
# ---------------------------------------

approval = df["LoanStatus"].value_counts()

plt.figure(figsize=(5,4))

approval.plot(kind="bar")

plt.title("Loan Approval Count")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applicants")

plt.show()

# ---------------------------------------
# Step 9 : Scatter Plot
# ---------------------------------------

colors=[]

for status in df["LoanStatus"]:
    if status=="Approved":
        colors.append("green")
    else:
        colors.append("red")

plt.figure(figsize=(6,4))

plt.scatter(df["ApplicantIncome"],
            df["LoanAmount"],
            c=colors)

plt.title("Income vs Loan Amount")
plt.xlabel("Applicant Income")
plt.ylabel("Loan Amount")

plt.show()

# ---------------------------------------
# Step 10 : Pie Chart
# ---------------------------------------

plt.figure(figsize=(5,5))

approval.plot(kind="pie",
              autopct="%1.1f%%")

plt.title("Loan Approval Percentage")
plt.ylabel("")

plt.show()

# ---------------------------------------
# Step 11 : Simple Loan Prediction
# ---------------------------------------

print("\n==============================")
print("LOAN PREDICTION")
print("==============================")

income = int(input("Enter Applicant Income : "))
loan = int(input("Enter Loan Amount : "))
credit = int(input("Enter Credit History (1 or 0): "))

if income >= 25000 and credit == 1 and loan <= 150:
    print("\nLoan Prediction : APPROVED")
else:
    print("\nLoan Prediction : REJECTED")

print("\nProject Completed Successfully.")
```

```python
import pandas as pd
data1={"name":["pavani","anitha","chinna"],"roll no":[101,102,103]}
data2={"attendance":[80,70,95]}
data3={"dbms":[85,80,90],"DAE":[75,65,84],"C++":[80,93,64]}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
df3=pd.DataFrame(data3)
df=pd.concat([df1,df2,df3],axis=1)
df["attendance_status"]=df["attendance"].apply(lambda x:"eligible for exam" if x>75 else "detained")
df["dbms_status"]=df["dbms"].apply(lambda x:"pass" if x>=75 else "fail")
df["DAE_status"]=df["DAE"].apply(lambda x:"pass" if x>=75 else "fail")
df["C++_status"]=df["C++"].apply(lambda x:"pass" if x>=75 else "fail")
print(df)
```

```python
#vertical stacking
import numpy as np
arr1 = np.array([1, 2, 3])
arr2 = np.array([[4, 5, 6],[5,7,9]])
varr = np.vstack((arr1, arr2))
print(varr)
print(varr.ndim)
```

```python
#horizontal stacking
import numpy as np
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
varr = np.hstack((arr1, arr2))
print(varr)
print(varr.ndim)
```
